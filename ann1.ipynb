```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#DEfine activation functions
def sigmoid(x):
    return 1/(1+np.exp(-x))

def tanh(x):
    return (np.exp(x)-np.exp(-x))/(np.exp(x)+np.exp(-x)) 

# ReLU
def relu(x):
    return np.maximum(0, x)

# Leaky ReLU
def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)
```

```python
#generate range of values
x=np.linspace(-5,5,100)
```

```python
y_sig=sigmoid(x)
y_t=tanh(x)
y_relu=relu(x)
y_leakyR=leaky_relu(x)
```

```python
plt.figure(figsize=(8,6))
plt.plot(x,y_sig,label='Sigmoid',color='blue')
plt.title('Sigmoid Activvation Function')
plt.xlabel('Input(x)')
plt.ylabel('Output(f(x))')
plt.axhline(0,color='red',linestyle='--',linewidth=1)
plt.axvline(0,color='red',linestyle='--',linewidth=1)
plt.legend()
plt.grid(True)
plt.show
```

```python
plt.figure(figsize=(8,6))
plt.plot(x,y_t,label='tanh',color='blue')
plt.title('tanh Function')
plt.xlabel('Input(x)')
plt.ylabel('Output(f(x))')
plt.axhline(0,color='red',linestyle='--',linewidth=1)
plt.axvline(0,color='red',linestyle='--',linewidth=1)
plt.legend()
plt.grid(True)
plt.show
```

```python
plt.figure(figsize=(8,6))
plt.plot(x,y_relu,label='relu',color='blue')
plt.title('relu Function')
plt.xlabel('Input(x)')
plt.ylabel('Output(f(x))')
plt.axhline(0,color='red',linestyle='--',linewidth=1)
plt.axvline(0,color='red',linestyle='--',linewidth=1)
plt.legend()
plt.grid(True)
plt.show
```

```python
plt.figure(figsize=(8,6))
plt.plot(x,y_leakyR,label='leaky_relu',color='blue')
plt.title('Leaky relu Function')
plt.xlabel('Input(x)')
plt.ylabel('Output(f(x))')
plt.axhline(0,color='red',linestyle='--',linewidth=1)
plt.axvline(0,color='red',linestyle='--',linewidth=1)
plt.legend()
plt.grid(True)
plt.show
```
