```python
import numpy as np
```

```python
class HopfieldNetwork:
    def __init__(self, size):
        self.size = size
        self.weights = np.zeros((size, size), dtype=float)

    def training(self, patterns):
        self.weights = np.zeros((self.size, self.size), dtype=float)

        for p in patterns:
            p = np.array(p, dtype=float)
            self.weights += np.outer(p, p)

        np.fill_diagonal(self.weights, 0)
        self.weights /= self.size

    def recall(self, pattern, steps=10):
        state = np.array(pattern, dtype=float)

        for _ in range(steps):
            for i in range(self.size):
                raw = np.dot(self.weights[i], state)
                state[i] = 1 if raw >= 0 else -1

        return state

    def energy(self, state):
        state = np.array(state, dtype=float)
        return -0.5 * np.dot(state.T, np.dot(self.weights, state))
```

```python
hopfield=HopfieldNetwork(3)
pattern=[
    [1,-1,1],
    [1,1,1]
]
hopfield.training(pattern)

test=[1,1,-1]

result=hopfield.recall(test)

energy=hopfield.energy(result)

result,energy
```
