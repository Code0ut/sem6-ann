```python
import numpy as np
```

```python
inp=np.array([[0,0],
     [0,1],
     [1,0],
     [1,1]])
```

```python
op=np.array([0,1,1,0])
op=op.reshape(-1,1)
print(op)
```

```python
np.random.seed(42)
```

```python
ip_neurons=2
hidden_neurons=3
op_neurons=1
```

```python
W1=np.random.uniform(-1,1,size=(ip_neurons,hidden_neurons))
b1=np.random.uniform(-1,1,size=(1,hidden_neurons))
```

```python
print(W1)
```

```python
print(b1)
```

```python
W2=np.random.uniform(-1,1,size=(hidden_neurons,op_neurons))
b2=np.random.uniform(-1,1,size=(1,op_neurons))
```

```python
print(W2)
```

```python
print(b2)
```

```python
def sigmoid(x):
    return 1/(1+np.exp(-x))
```

```python
def sigmoid_d(x):
    return x*(1-x)
```

```python
lr=0.1
epochs=1000
```

```python
for epoch in range(epochs):

    hidden_ip=np.dot(inp,W1)+b1
    hidden_op=sigmoid(hidden_ip)

    final_ip=np.dot(hidden_op,W2)+b2
    final_op=sigmoid(final_ip)

    error=op-final_op

    d_op=error*sigmoid_d(final_op)
    
    hidden_error = np.dot(d_op, W2.T) 

    d_hidden=hidden_error*sigmoid_d(hidden_op)


    W2+=np.dot(hidden_op.T,d_op)*lr
    b2+=np.sum(d_op,axis=0,keepdims=True)*lr

    W1+=np.dot(inp.T,d_hidden)*lr
    b1+=np.sum(d_hidden,axis=0,keepdims=True)*lr
    if epoch % 100 == 0:
        print("Loss:", np.mean(np.square(error)))
```
