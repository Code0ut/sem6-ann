```python
import random

# Input (ASCII digits split)
inp = []
for i in range(48, 58):
    inp.append([int(d) for d in str(i)])

# Target: Even = 1, Odd = -1
op = [1 if i % 2 == 0 else -1 for i in range(10)]

# REQUIRED activation
def activation(y_in, theta=0):
    if y_in > theta:
        return 1
    elif y_in < -theta:
        return -1
    else:
        return 0

# Training
def perceptron_train(n_epochs, inp, op, weights, bias, lr):
    for ep in range(n_epochs):
        for i in range(len(inp)):
            x = inp[i]
            target = op[i]

            y_in = sum(weights[j]*x[j] for j in range(len(x))) + bias
            y = activation(y_in)

            if y != target:   
                for j in range(len(weights)):
                    weights[j] += lr * target * x[j]
                bias += lr * target

    return weights, bias

# Initialize
weights = [random.uniform(-1,1) for _ in range(len(inp[0]))]
bias = random.uniform(-1,1)

# Train
weights, bias = perceptron_train(20, inp, op, weights, bias, 0.1)

# Test
pred = []
for i in range(len(inp)):
    y_in = sum(weights[j]*inp[i][j] for j in range(len(inp[i]))) + bias
    pred.append(activation(y_in))

print("Predictions:", pred)
```
